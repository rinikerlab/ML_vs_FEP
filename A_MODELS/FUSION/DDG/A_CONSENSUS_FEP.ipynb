```python
""""
Copyright (C) 2025 ETH Zurich, Stefan J.P. Chrin, Gregory Landrum, Sereina Riniker
"""

import json
import pandas as pd

# path
path_MET = "../../Z_META/DDG/RANDOM/ddG_ridge_LOO_pred&error.json"
path_OFF = "../../FEP/DDG/OFF/OFF_RANDOM_ddG_FROM_FEP_GRAPH_pred&error.json"

# load json
with open(path_MET, "r") as f:
     met_predictions= json.load(f)
     
with open(path_OFF, "r") as f:
     OFF_predictions= json.load(f)
    
```

```python
import json
import copy

# deep copies
avg_predictions = copy.deepcopy(met_predictions)
max_predictions = copy.deepcopy(met_predictions)
min_predictions = copy.deepcopy(met_predictions)

# iterate splits
for i, split in enumerate(avg_predictions):
    for target in split:
        for pair in split[target]:
            # get preds
            met_pred = met_predictions[i][target][pair]["prediction"]
            off_pred = OFF_predictions[i][target][pair]["prediction"]
            exp_val = met_predictions[i][target][pair]["experimental"]

            # compute
            avg_pred = (met_pred + off_pred) / 2.0
            max_pred = max(met_pred, off_pred)
            min_pred = min(met_pred, off_pred)

            # assign avg
            avg_predictions[i][target][pair]["prediction"] = avg_pred
            avg_predictions[i][target][pair]["experimental"] = exp_val

            # assign max
            max_predictions[i][target][pair]["prediction"] = max_pred
            max_predictions[i][target][pair]["experimental"] = exp_val

            # assign min
            min_predictions[i][target][pair]["prediction"] = min_pred
            min_predictions[i][target][pair]["experimental"] = exp_val

# save
with open("A_OUTPUT/ddG_ridge_LOO_pred&error_AVG_MET_OFF.json", "w") as f:
    json.dump(avg_predictions, f, indent=4)

with open("A_OUTPUT/ddG_ridge_LOO_pred&error_MAX_MET_OFF.json", "w") as f:
    json.dump(max_predictions, f, indent=4)

with open("A_OUTPUT/ddG_ridge_LOO_pred&error_MIN_MET_OFF.json", "w") as f:
    json.dump(min_predictions, f, indent=4)
```
