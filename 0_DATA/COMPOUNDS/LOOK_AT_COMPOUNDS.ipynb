```python
""""
Copyright (C) 2025 ETH Zurich, Stefan J.P. Chrin, Gregory Landrum, Sereina Riniker
"""


# ---------- params ----------
N_COLS = 5
IMG_W, IMG_H = 180, 180
PADDING_FRAC = 0.06
PER_PAGE = 25
FONT_SIZE = 12
USE_SVG = True          
REMOVE_HS = True        
KEKULIZE = True         
SAVE_FILES = False      
DPI = 150
OUT_DIR = "_renders"
# ---------------------------

from pathlib import Path
from typing import Iterable, List, Tuple
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.Draw import rdMolDraw2D
from rdkit.Chem import AllChem
from IPython.display import display, SVG

def _chunk(it: Iterable, n: int):
    """input: iterable, n>0; behavior: yields chunks of size n; output: generator"""
    it = list(it)
    for i in range(0, len(it), n):
        yield it[i:i+n]

def _read_sdf_mols(sdf_path: Path) -> List[Chem.Mol]:
    """input: .sdf path; behavior: reads, keeps valid; output: list[Mol]"""
    suppl = Chem.SDMolSupplier(str(sdf_path), removeHs=False, sanitize=True)
    return [m for m in suppl if m is not None]

def _prep_for_drawing(m: Chem.Mol) -> Chem.Mol:
    """input: Mol; behavior: copy, 2D coords, remove Hs, kekulize; output: Mol"""
    m2 = Chem.Mol(m)
    if REMOVE_HS:
        m2 = Chem.RemoveHs(m2)
    if KEKULIZE:
        try:
            Chem.Kekulize(m2, clearAromaticFlags=True)
        except Exception:
            pass
    AllChem.Compute2DCoords(m2)
    return m2

def _grid_image(mols: List[Chem.Mol], legends: List[str]):
    """input: mols+legends; behavior: build grid; output: SVG str or PIL Image"""
    mols = [_prep_for_drawing(m) for m in mols if m is not None]
    legends = legends[:len(mols)]

    dops = rdMolDraw2D.MolDrawOptions()
    dops.padding = float(PADDING_FRAC)
    dops.legendFontSize = int(FONT_SIZE)

    img = Draw.MolsToGridImage(
        mols,
        molsPerRow=int(N_COLS),
        subImgSize=(int(IMG_W), int(IMG_H)),
        legends=legends,
        useSVG=bool(USE_SVG),
        drawOptions=dops,
        returnPNG=False,
    )
    return img

def _mols_and_legends_from_dir(target_dir: Path) -> Tuple[List[Chem.Mol], List[str]]:
    """input: dir with .sdf; behavior: combine; output: (mols, legends)"""
    mols, legends = [], []
    for sdf in sorted(p for p in target_dir.iterdir() if p.suffix.lower()==".sdf"):
        ms = _read_sdf_mols(sdf)
        mols.extend(ms)
        for i, m in enumerate(ms):
            name = m.GetProp("_Name") if m.HasProp("_Name") else f"{sdf.stem}_{i+1}"
            legends.append(name)
    return mols, legends

def _save_image(img, out_path: Path):
    """input: img,str path; behavior: save svg/png; output: None"""
    out_path.parent.mkdir(parents=True, exist_ok=True)
    if isinstance(img, str):           # SVG
        out_path.write_text(img, encoding="utf-8")
    else:                              # PIL
        img.save(str(out_path), dpi=(DPI, DPI))

def render_target_dir_inline(target_dir: Path, out_dir: Path):
    """input: target_dir, out_dir; behavior: show inline; (opt) save; output: None"""
    mols, legends = _mols_and_legends_from_dir(target_dir)
    if not mols:
        print(f"No molecules in {target_dir}")
        return

    page_idx = 1
    for cm, cl in zip(_chunk(mols, PER_PAGE), _chunk(legends, PER_PAGE)):
        img = _grid_image(cm, cl)

        # --- single display ---
        if isinstance(img, str):      
            display(SVG(img))
        else:                         
            display(img)
        # ---------------------------

        if SAVE_FILES:
            tgt = target_dir.name
            ext = "svg" if USE_SVG else "png"
            _save_image(img, out_dir / tgt / f"{tgt}_page{page_idx}.{ext}")
        page_idx += 1

def render_targets_inline(base_dir: str):
    """input: base_dir with subfolders; behavior: render each inline; output: None"""
    base = Path(base_dir).expanduser().resolve()
    out_root = base.parent / OUT_DIR
    for target_dir in sorted(p for p in base.iterdir() if p.is_dir()):
        print(f"== {target_dir.name} ==")
        render_target_dir_inline(target_dir, out_root)

# run
base_dir = "PER_TARGET_SDF"
render_targets_inline(base_dir)
```
